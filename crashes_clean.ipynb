```python
import pandas as pd

url = "https://raw.githubusercontent.com/SujalS14/Road-Traffic-Accidents/refs/heads/main/Road%20Traffic%20Accident%20Dataset%20of%20Addis%20Ababa%20City/RTA%20Dataset.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
clean_data = df.drop_duplicates()

cols = clean_data.select_dtypes(include=['object', 'category']).columns
clean_data[cols] = clean_data[cols].fillna('Unknown')

print(f"Original rows: {len(df)}, Rows after dropping duplicates: {len(clean_data)}")

clean_data['Time'] = pd.to_datetime(clean_data['Time'], format='%H:%M:%S', errors='coerce')
clean_data['hour'] = clean_data['Time'].dt.hour

clean_data['is_high_severity'] = clean_data['Accident_severity'].apply(
    lambda x: 1 if str(x).strip().lower() in ['serious injury', 'fatal injury'] else 0
)

display(clean_data[['Time', 'hour', 'Accident_severity', 'is_high_severity']].head())

total_number_crashes = len(clean_data)

share_high_severity = clean_data['is_high_severity'].mean()

top_10_causes = clean_data['Cause_of_accident'].value_counts().head(10)
```

```python
from google.colab import files

clean_data.to_csv("crashes_clean.csv", index=False)
files.download("crashes_clean.csv")
```
