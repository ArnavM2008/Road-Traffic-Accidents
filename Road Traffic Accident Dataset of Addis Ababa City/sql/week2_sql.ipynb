```python
import pandas as pd
import sqlite3

df = pd.read_csv('crashes_clean.csv')
conn = sqlite3.connect(':memory:')
df.to_sql('data', conn, index=False, if_exists='replace')

def q(sql):
    return pd.read_sql_query(sql, conn)

print(df.shape)
```

```python
# Q1: Which light conditions have the highest share of serious or fatal crashes?
q("""
SELECT Light_conditions,
       COUNT(*) AS crashes,
       AVG(is_high_severity) AS high_severity_share
FROM data
GROUP BY Light_conditions
ORDER BY high_severity_share DESC
""")
```

```python
# Q2: Which weather conditions have the fewest recorded crashes (data gap)?
# If this dataset is used for training, the model will have the fewest examples for these conditions.
q("""
SELECT Weather_conditions,
       COUNT(*) AS crashes
FROM data
GROUP BY Weather_conditions
ORDER BY crashes ASC
""")
```

```python
# Q3: Does weather affect crash severity?
q("""
SELECT Weather_conditions,
       COUNT(*) AS crashes,
       AVG(is_high_severity) AS high_severity_share
FROM data
GROUP BY Weather_conditions
ORDER BY high_severity_share DESC
""")
```

```python
# Q4: Which junction types are the riskiest?
q("""
SELECT Types_of_Junction,
       COUNT(*) AS crashes,
       AVG(is_high_severity) AS high_severity_share
FROM data
GROUP BY Types_of_Junction
ORDER BY high_severity_share DESC
""")
```

```python
# Q5: What time of day has the most crashes?
q("""
SELECT hour,
       COUNT(*) AS crashes
FROM data
GROUP BY hour
ORDER BY crashes DESC
LIMIT 5
""")
```
